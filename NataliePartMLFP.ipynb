# 1.2

```python
plt.figure(figsize=(12,6))
df_sorted = df.sort_values("avg_price", ascending=False)

sns.barplot(
    data=df_sorted,
    x="brand",
    y="avg_price",
    palette="Blues_r"
)

plt.title("Average Air Fryer Price by Brand", fontsize=16, weight='bold')
plt.xlabel("Brand", fontsize=12)
plt.ylabel("Average Price ($)", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))
df_sorted = df.sort_values("avg_rating", ascending=False)

sns.barplot(
    data=df_sorted,
    x="brand",
    y="avg_rating",
    palette="Blues_r"
)

plt.title("Average Air Fryer Rating by Brand", fontsize=16, weight='bold')
plt.xlabel("Brand", fontsize=12)
plt.ylabel("Average Rating", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))
df_sorted = df.sort_values("brand_share", ascending=False)

sns.barplot(
    data=df_sorted,
    x="brand",
    y="brand_share",
    palette="Blues_r"
)

plt.title("Average Air Fryer Market Share by Brand", fontsize=16, weight='bold')
plt.xlabel("Brand", fontsize=12)
plt.ylabel("Market Share", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
